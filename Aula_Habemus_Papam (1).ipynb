```python
import cv2 as cv
import numpy as np

imagem = cv.imread("imagens/wirebond_mask.tif")

imagem = cv.blur(imagem, (3, 3))
cv.imshow("Imagem original", imagem)

laplace = cv.Laplacian(imagem, cv,CV_64f)
cv.imshow("Laplace [com valores invertidos]", laplace)

laplace = np.uint8(np.absolute(laplace))
cv.imshow("Laplace [com valores ajustados", laplace)

cv.waitKey(0)
cv.destroyAllWindows()
```
