```python
import cv2 as cv 
import numpy as np 

img = cv.imread("imagens/lena_ruido.bmp")
img_media = cv.blur(img, (5, 5))
img_mediana = cv.medianBlur(img, 5)

cv.imshow("Imagem Original", img)
cv.imshow("Imagem Filtro Media", img_media)
cv.imshow("Imagem Filtro Mediana", img_mediana)

cv.waitKey(0)
cv.DestroyAllWindows()
```

```python
import cv2 as cv

img = cv.imread("imagens/lena_original.png")
img_borrada = cv.imread("imagens/lena_borrada.png")

img_mask = cv.subtract(img, img_borrada)

cv.imshow("Imagem Original", img)
cv.imshow("Imagem Borrada", img_borrada)

imagem = cv.multiply(2, img_mask)
cv.imshow("Imagem Ponderada", imagem)

img_agucada = cv.add(img, imagem)
cv.imshow("Imagem Aguçada", img_agucada)

cv.waitKey(0)
cv.DestroyAllWindows()
```
