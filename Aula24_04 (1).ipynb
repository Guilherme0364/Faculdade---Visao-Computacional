```python
import cv2 as cv
import numpy as np

img = cv.imread("imagens/estrada.jpg")

# O underline é irrelevante, precisamos dele somente porque a função pede 3 parâmetros
h, w, _ = img.shape

escala = 40
img = cv.resize(img, dsize = (round(escala / 100.0 * w), round(escala / 100.0 * h)))

img_suavizada = np.vstack([
    np.hstack([img, cv.blur(img, (3, 3)), cv.blur(img, (5, 5))]),
    np.hstack([cv.blur(img, (7, 7)), cv.blur(img, (9, 9)), cv.blur(img, (11, 11))]),
])

cv.imshow("Imagens suavizadas", img_suavizada)

cv.waitKey(0)
cv.destroyAllWindows()
```
