```python
import cv2 as cv
import  numpy as np 

img_retangulo = np.ones((400, 400), dtype = "uint8")

cv.rectangle(img_retangulo, (50, 50), (300, 300), (255, 255, 255), -1)

# Círculo

img_circulo = np.ones((400, 400), dtype = "uint8")

cv.circle(img_circulo, (300, 300), 70, (255, 255, 255) , -1)

op_and = cv.bitwise_and(img_retangulo, img_circulo)
op_or = cv.bitwise_or(img_retangulo, img_circulo)
op_not_ret = cv.bitwise_not(img_retangulo)
op_not_cir = cv.bitwise_not(img_circulo)

cv.imshow("Retângulo", img_retangulo)
cv.imshow("Círculo", img_circulo)
cv.imshow("AND", op_and)
cv.imshow("OR", op_or)
cv.imshow("NOT - Retângulo", op_not_ret)
cv.imshow("NOT - Círculo", op_not_cir)

cv.waitKey(0)
cv.destroyAllWindows()
```
