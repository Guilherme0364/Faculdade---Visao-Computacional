```python
import cv2  as cv
```

```python
img = cv.imread("lena.png", cv.IMREAD_GRAYSCALE)

cv.imshow("Imagem da Lena", img)

cv.waitKey(2000)
cv.destroyAllWindows()
```

```python
img = cv.imread("lena.png", cv.IMREAD_GRAYSCALE)

cv.imshow("Imagem da Lena", img)

cv.imwrite("nova_imagem.jpeg", img)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
img = cv.imread("lena.png", cv.IMREAD_GRAYSCALE)

cv.imshow("Imagem da Lena", img)

k = cv.waitKey(0)

if k == 27:
    cv.destroyAllWindows()                 # Espera pela tecla ESC
elif k == ord('s'):                         # Espera pela tecla 's'
    cv.imwrite("nova_imagem01.png", img)
    cv.destroyAllWindows()
```

```python
img = cv.imread("lena.png", cv.IMREAD_READCOLOR)

cv.imshow("Imagem da Lena", img)

print(img.shape)

cv.waitKey(0)
cv.destroyAllWindows()
```
